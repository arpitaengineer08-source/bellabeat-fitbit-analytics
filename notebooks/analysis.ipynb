# Bellabeat / Fitbit — Starter EDA
Loads the cleaned CSVs and gives a first look at the data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

CLEAN_DIR = os.path.join('..', 'data', 'cleaned')
sns.set_style('whitegrid')
```

```python
daily = pd.read_csv(os.path.join(CLEAN_DIR, 'dailyActivity_merged.csv'), parse_dates=['ActivityDate'])
sleep = pd.read_csv(os.path.join(CLEAN_DIR, 'sleepDay_merged.csv'), parse_dates=['SleepDay'])
weight = pd.read_csv(os.path.join(CLEAN_DIR, 'weightLogInfo_merged.csv'), parse_dates=['Date'])

daily.head()
```

```python
# Quick shape / sanity check
print('daily activity :', daily.shape)
print('sleep          :', sleep.shape)
print('weight         :', weight.shape)
print('unique users (daily):', daily['Id'].nunique())
```

```python
# Steps vs. Calories
fig, ax = plt.subplots(figsize=(7, 5))
sns.scatterplot(data=daily, x='TotalSteps', y='Calories', ax=ax)
ax.set_title('Total Steps vs. Calories Burned')
plt.savefig('../outputs/steps_vs_calories.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Activity minutes breakdown, averaged across users
activity_cols = ['VeryActiveMinutes', 'FairlyActiveMinutes', 'LightlyActiveMinutes', 'SedentaryMinutes']
avg_minutes = daily[activity_cols].mean()

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(avg_minutes, labels=avg_minutes.index, autopct='%1.1f%%')
ax.set_title('Average Daily Minutes by Activity Level')
plt.savefig('../outputs/activity_breakdown.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Merge sleep and daily activity to compare sleep vs. activity
merged = pd.merge(
    daily, sleep,
    left_on=['Id', 'ActivityDate'], right_on=['Id', 'SleepDay'],
    how='inner'
)
print(merged.shape)
merged[['TotalSteps', 'TotalMinutesAsleep', 'TotalTimeInBed']].corr()
```
